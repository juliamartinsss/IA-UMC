19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1, 10, 20)  #gera o primeiro array com 20 números aleatórios entre 1 e 9
array2 = np.random.randint(1, 10, 20)  #gera o segundo array com 20 números aleatórios entre 1 e 9

produtoEscalar = np.dot(array1, array2)  #calcula o produto escalar entre os dois arrays
comprimentoA = np.sqrt(np.dot(array1, array1))  #calcula o comprimento do primeiro vetor
comprimentoB = np.sqrt(np.dot(array2, array2))  #calcula o comprimento do segundo vetor

anguloCos = produtoEscalar / (comprimentoA * comprimentoB)  #calcula o cosseno do ângulo (produto escalar normalizado)
angulo_rad = np.arccos(anguloCos)  #obtém o ângulo em radianos
angulo_graus = np.degrees(angulo_rad)  #converte o ângulo para graus

print(f'Produto Escalar: {produtoEscalar}')  #exibe o produto escalar
print(f'Ângulo em graus: {angulo_graus:.2f}°')  #exibe o ângulo em graus com duas casas decimais
```

Explicação: Dois arrays unidimensionais foram criados com números inteiros aleatórios de 1 a 9, cada um com 20 elementos. O produto escalar foi calculado com np.dot. Os comprimentos dos vetores foram obtidos pela raiz quadrada do produto escalar de cada vetor por ele mesmo. O cosseno do ângulo foi encontrado dividindo o produto escalar pelo produto dos comprimentos. Por fim, np.arccos converte o cosseno em ângulo radiano e np.degrees transforma esse valor para graus, exibindo o resultado com duas casas decimais.

1. Calcule a integral da função f(x) = x^2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy import integrate  #importa a função integrate da biblioteca scipy

def f(x):  #define a função a ser integrada
    return (x**2) * np.sin(x)
resultado, erro = integrate.quad(f, 0, np.pi)  #calcula a integral definida de 0 até pi

print('Resultado da integral:', resultado)  #exibe o resultado da integral
print('Erro estimado:', erro)  #exibe o erro estimado do cálculo
```

Explicação: Uma função f(x) foi definida para representar a expressão x² · sen(x), utilizando np.sin para o cálculo do seno. A função integrate.quad() da biblioteca SciPy foi então empregada para calcular a integral definida no intervalo de 0 a π, recebendo a função e os limites como argumentos. O resultado da integral e uma estimativa de erro são armazenados nas variáveis resultado e erro, exibidos ao final.


2.  Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.

```python
import numpy as np
from scipy.integrate import solve_ivp  #importa a função solve_ivp do módulo integrate do scipy

def f(x, y):  #define a equação diferencial dy/dx = x - y
    return x - y

intervalo = [0, 10]  #define o intervalo de integração de 0 a 10
y0 = [1]  #define a condição inicial y(0) = 1
x_pontos = np.linspace(0, 10, 100)  #gera 100 pontos igualmente espaçados entre 0 e 10
solucao = solve_ivp(f, intervalo, y0, t_eval=x_pontos)  #resolve a equação diferencial nos pontos definidos

print('Valores de x:', solucao.t)  #exibe os valores de x
print('Soluções para y:', solucao.y)  #exibe as soluções para y
```

Explicação: Uma função foi criada para representar a equação diferencial dy/dx = x - y. O intervalo de 0 a 10 e a condição inicial y(0) = 1 foram definidos. A função linspace() gera 100 pontos espaçados uniformemente no intervalo. solve_ivp() resolve a equação diferencial, recebendo a função, o intervalo, a condição inicial e os pontos de avaliação. Por fim, são exibidos os valores de x e as respectivas soluções de y.



3. Encontre as raízes da função f(x) = x^4 - 3x^3 + 2 utilizando SciPy.


```python
from scipy.optimize import fsolve  #importa a função fsolve do módulo optimize do scipy

def f(x):  #define a função cujas raízes serão encontradas
    return x**4 - 3*x**3 + 2

raiz1 = fsolve(f, 1.0)  #encontra uma raiz próxima de x = 1
raiz2 = fsolve(f, 3.0)  #encontra outra raiz próxima de x = 3

print(f'Raiz encontrada próxima de 1: {raiz1}')  #exibe a raiz encontrada próxima de 1
print(f'Raiz encontrada próxima de 3: {raiz2}')  #exibe a raiz encontrada próxima de 3
```

Explicação: Uma função foi definida para representar a equação x⁴ - 3x³ + 2. A função fsolve() do módulo scipy.optimize foi utilizada para localizar as raízes, recebendo a função e um palpite inicial. A primeira raiz foi buscada a partir de x = 1.0 e a segunda a partir de x = 3.0. Os resultados foram armazenados e exibidos.

4. Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1.

```python
from scipy.optimize import minimize  #importa a função minimize do módulo optimize do scipy

def f_negativa(x):  #define a função negativa para transformar a maximização em minimização
    return -(x**2 - 4*x - 1)

resultado = minimize(f_negativa, 0)  #encontra o mínimo da função negativa partindo de x = 0
ponto_maximo = resultado.x[0]  #extrai o valor de x do array retornado
valor_maximo = -resultado.fun  #inverte o sinal para obter o valor máximo da função original

print(f'O ponto de máximo encontrado é x = {ponto_maximo:.2f}')  #exibe o ponto de máximo
print(f'O valor máximo de f(x) é {valor_maximo:.2f}')  #exibe o valor máximo da função
```

Explicação: Uma função foi criada para representar a forma negativa da expressão original, permitindo transformar o problema de maximização em um de minimização. A função minimize() foi aplicada a partir do palpite inicial x = 0. O resultado contém o ponto ótimo e o valor da função nesse ponto. O valor de x é extraído do array resultado.x, e o valor máximo é obtido invertendo o sinal de resultado.fun.

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).


```python
import numpy as np
from scipy.fft import fft  #importa a função fft do módulo fft do scipy

t = np.linspace(0, 1, 500)  #cria um intervalo de tempo de 0 a 1 segundo com 500 amostras
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)  #define o sinal composto pela soma de seno e cosseno
frequencias = fft(sinal)  #calcula a transformada de Fourier do sinal

print('Transformada de Fourier (primeiros 5 valores):')  #exibe mensagem indicando os primeiros valores
print(frequencias[:5])  #exibe os primeiros 5 valores da transformada
```

Explicação: Um vetor de tempo foi criado com np.linspace, gerando 500 pontos entre 0 e 1 segundo. O sinal foi construído pela soma de uma onda senoidal de frequência 1 Hz e uma cossenoidal de 2 Hz. A função fft() do módulo scipy.fft calcula a Transformada Rápida de Fourier, convertendo o sinal do domínio do tempo para o domínio da frequência. O resultado é armazenado em frequencias, e apenas os cinco primeiros valores são exibidos.

6. Encontre a matriz inversa de uma matriz 5x5 de números aleatorios.

```python
import numpy as np
from scipy import linalg  #importa o módulo linalg do scipy

matriz = np.random.randint(1, 50, size=(5, 5))  #gera uma matriz 5x5 com números aleatórios entre 1 e 49
matriz_inversa = linalg.inv(matriz)  #calcula a matriz inversa

print('Matriz Original (5x5):\n', matriz)  #exibe a matriz original
print('\nMatriz Inversa:\n', matriz_inversa)  #exibe a matriz inversa

identidade = np.dot(matriz, matriz_inversa)  #verifica o produto entre a matriz original e a inversa

print('\nVerificação (Matriz Identidade):\n', np.round(identidade, 2))  #exibe a matriz identidade arredondada
```

Explicação: Uma matriz 5x5 foi gerada com números inteiros aleatórios entre 1 e 49. A função linalg.inv() do SciPy foi utilizada para calcular a matriz inversa. O produto entre a matriz original e sua inversa foi calculado com np.dot, e o resultado deve ser aproximadamente uma matriz identidade; o arredondamento com np.round facilita a visualização da verificação.

7. Use a funções de interpolação cúbica de SciPy para interpolar os dados
(0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

```python
import numpy as np
from scipy.interpolate import interp1d  #importa a função interp1d do módulo interpolate do scipy

x = np.array([0, 1, 2, 3, 4])  #define os pontos conhecidos de x
y = np.array([0, 2, 3, 5, 4])  #define os pontos correspondentes de y

f_cubica = interp1d(x, y, kind='cubic')  #cria uma função de interpolação cúbica
x_novos = np.linspace(0, 4, 7)  #gera 7 novos pontos entre 0 e 4
y_novos = f_cubica(x_novos)  #calcula os valores interpolados para os novos pontos

print('Novos pontos x:', x_novos)  #exibe os novos pontos de x
print('Valores interpolados y:', y_novos)  #exibe os valores interpolados de y
```

Explicação: Foram definidos arrays x e y com os cinco pontos fornecidos no enunciado. A função interp1d com kind='cubic' cria um interpolador que usa um polinômio de terceiro grau. np.linspace gera sete novos pontos uniformemente distribuídos entre 0 e 4, e o interpolador calcula os valores correspondentes de y, que são então exibidos.

8. Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6.

```python
import numpy as np
from scipy import linalg  #importa o módulo linalg do scipy

A = np.array([[3, 2], [1, 4]])  #define a matriz dos coeficientes das equações
B = np.array([5, 6])  #define o vetor dos termos independentes
solucao = linalg.solve(A, B)  #resolve o sistema linear

print(f'Solução [x, y]: {solucao}')  #exibe a solução encontrada
```

Explicação: Uma matriz A foi criada com os coeficientes das variáveis (3, 2 e 1, 4) e um vetor B com os resultados das equações (5 e 6). A função linalg.solve() resolve o sistema linear, retornando os valores de x e y que satisfazem ambas as equações.

9. Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).


```python
import numpy as np
from scipy.optimize import curve_fit  #importa a função curve_fit do módulo optimize do scipy

x_dados = np.array([1, 2, 3, 4, 5])  #define os pontos de dados de x
y_dados = np.array([2, 4, 5, 7, 8])  #define os pontos de dados de y

def curva_quadratica(x, a, b, c):  #define a função modelo (curva quadrática/parábola)
    return a * x**2 + b * x + c
parametros, _ = curve_fit(curva_quadratica, x_dados, y_dados)  #ajusta a curva aos pontos

print('Parâmetros ajustados [a, b, c]:')  #exibe mensagem indicando os parâmetros
print(parametros)  #exibe os parâmetros ajustados
```

Explicação: Um array x_dados e um array y_dados foram criados contendo os 5 pontos correspondentes fornecidos no enunciado. A função curva_quadratica define o modelo de parábola. O curve_fit ajusta os parâmetros a, b e c para minimizar o erro entre a curva e os pontos. Por fim, os parâmetros ajustados são exibidos.


10. Calcule a integral dupla da funções f(x, y) = xy + x^2 no intervalo [0, 2]
para x e [0, 1] para y.


```python
from scipy.integrate import dblquad  #importa a função dblquad do módulo integrate do scipy

def f(y, x):  #define a função a ser integrada (note a ordem y, x exigida pelo dblquad)
    return x * y + x**2

# Limites de x (externos): 0 a 2
# Limites de y (internos): 0 a 1
resultado, erro = dblquad(f, 0, 2, 0, 1)  #calcula a integral dupla

print(f'Resultado da integral dupla: {resultado:.4f}')  #exibe o resultado com 4 casas decimais
print(f'Erro estimado: {erro}')  #exibe o erro estimado
```

Explicação: A função f(y, x) representa xy + x². O dblquad() calcula a integral dupla, com os limites externos de x indo de 0 a 2 e os internos de y indo de 0 a 1. A ordem dos parâmetros na função é (y, x) porque dblquad espera essa sequência. O resultado e o erro estimado são exibidos.

16. Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0) = 1 e y'(0) = 0.

```python
from scipy.integrate import solve_ivp  #importa a função solve_ivp do módulo integrate do scipy

def equacao(x, y):  #define o sistema de equações de primeira ordem equivalente
    return [y[1], -y[0]]

resultado = solve_ivp(equacao, [0, 10], [1, 0])  #resolve a equação diferencial com as condições iniciais

print(resultado.y)  #exibe a solução da equação diferencial
```

Explicação: Para resolver uma equação de segunda ordem com solve_ivp, ela é transformada em um sistema de duas equações de primeira ordem. A função equacao retorna as derivadas de y e y'. O solver integra o sistema de 0 a 10, usando as condições iniciais y(0)=1 e y'(0)=0.

18. Gráfico de barras com cores diferentes para valores positivos e negativos.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago']  #define os meses
valores = np.array([12, -7, 15, -3, 8, -10, 5, -2])  #valores positivos e negativos

cores = ['green' if v >= 0 else 'red' for v in valores]  #define verde para positivos e vermelho para negativos

plt.bar(categorias, valores, color=cores, edgecolor='black', linewidth=0.7)  #cria o gráfico de barras colorido

plt.axhline(0, color='black', linewidth=1.2)  #linha horizontal no valor zero

plt.title('Resultados Mensais: Lucros e Prejuízos (Jan a Ago)')  #adiciona título
plt.xlabel('Meses')  #nomeia eixo x
plt.ylabel('Valores (em R$ Milhares)')  #nomeia eixo y
plt.grid(True, linestyle='--', alpha=0.5)  #adiciona grade tracejada

plt.show()  #exibe o gráfico
```

Explicação: A lista cores aplica verde para valores ≥ 0 e vermelho para negativos. A linha horizontal em zero (axhline) separa visualmente lucros e prejuízos, e a grade tracejada facilita a leitura.
